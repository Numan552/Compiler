```python
%%writefile op_parser.cpp
#include <iostream>
#include <stack>
#include <cctype>
using namespace std;

int precedence(char op) {
    if (op == '+' || op == '-') return 1;
    if (op == '*' || op == '/') return 2;
    return 0;
}

int applyOp(int a, int b, char op) {
    switch (op) {
        case '+': return a + b;
        case '-': return a - b;
        case '*': return a * b;
        case '/': return a / b;
    }
    return 0;
}

int evaluate(string s) {
    stack<int> values;
    stack<char> ops;

    for (int i = 0; i < s.length(); i++) {

        if (isdigit(s[i])) {
            values.push(s[i] - '0');
        }
        else {
            while (!ops.empty() &&
                   precedence(ops.top()) >= precedence(s[i])) {

                int b = values.top(); values.pop();
                int a = values.top(); values.pop();
                char op = ops.top(); ops.pop();

                values.push(applyOp(a, b, op));
            }
            ops.push(s[i]);
        }
    }

    while (!ops.empty()) {
        int b = values.top(); values.pop();
        int a = values.top(); values.pop();
        char op = ops.top(); ops.pop();

        values.push(applyOp(a, b, op));
    }

    return values.top();
}

int main() {
    string expr;

    // 👇 manual input
    cout << "Enter the arithmetic expression: ";
    cin >> expr;

    int result = evaluate(expr);

    cout << "Value is " << result << endl;

    return 0;
}
```
